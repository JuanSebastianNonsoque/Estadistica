Un comprador seleccionado al azar compra un rodillo y una lata de pintura. ¿Cuál es la probabilidad de que la pintura sea de látex?   

```julia
P_Latex = 0.75              
P_Semigloss = 1.0 - P_Latex     

P_Roller_given_Latex = 0.60         
P_Roller_given_Semigloss = 0.30     

P_Roller = (P_Roller_given_Latex * P_Latex) + (P_Roller_given_Semigloss * P_Semigloss)

P_Latex_given_Roller = (P_Roller_given_Latex * P_Latex) / P_Roller

P_Latex_given_Roller_red = round(P_Latex_given_Roller, digits=4)
P_Latex_given_Roller_porcentaje = round(P_Latex_given_Roller * 100, digits=2)

println("Dado que el comprador seleccionado al azar compró un rodillo y una lata de pintura, la probabilidad de que la pintura sea de látex es aproximadamente $P_Latex_given_Roller_red (o $P_Latex_given_Roller_porcentaje%).")
```

1.78: Denote por $A$, $B$ y $C$ los eventos de que un gran premio esté detrás de las puertas $A$, $B$ y $C$, respectivamente. Suponga que elegiste al azar una puerta, digamos la $A$. El presentador del programa abrió una puerta, digamos la $B$, y mostró que no había ningún premio detrás de ella. Ahora el presentador te ofrece la opción de permanecer en la puerta que elegiste ($A$) o cambiar a la puerta restante que no está abierta ($C$). Use la probabilidad para explicar si debes cambiar o no.

```julia
n_simulaciones = 100000
aciertos_quedarse = 0
aciertos_cambiar = 0

for i in 1:n_simulaciones
    premio = rand(1:3)       
    eleccion = 1           
    
    if eleccion == premio
        aciertos_quedarse += 1
    end
    
    if eleccion != premio
        aciertos_cambiar += 1
    end
end

prob_quedarse = round(aciertos_quedarse / n_simulaciones, digits=4)
prob_cambiar = round(aciertos_cambiar / n_simulaciones, digits=4)

println("Al analizar el problema de Monty Hall, la probabilidad de ganar si decides quedarte es de aproximadamente $prob_quedarse (33.33%), mientras que la probabilidad de ganar si decides cambiar de puerta es de aproximadamente $prob_cambiar (66.67%), por lo que se debe cambiar.")
```

1.82: Mediante la comparación de regiones apropiadas de diagramas de Venn, verifique que:

a: $(A \cap B) \cup (A \cap B') = A$

b: $A' \cap (B' \cup C) = (A' \cap B') \cup (A' \cap C)$

```julia
U = Set(1:1000)
A = Set(rand(1:1000, 400))
B = Set(rand(1:1000, 400))
C = Set(rand(1:1000, 400))

parte_a_izq = union(intersect(A, B), intersect(A, setdiff(U, B)))
valida_a = (parte_a_izq == A)

A_comp = setdiff(U, A)
B_comp = setdiff(U, B)
parte_b_izq = intersect(A_comp, union(B_comp, C))
parte_b_der = union(intersect(A_comp, B_comp), intersect(A_comp, C))
valida_b = (parte_b_izq == parte_b_der)

println("La verificación computacional de regiones de conjuntos confirma que la identidad (a) es valida y la ley distributiva (b) tambien es valida.")
```

Suponga que el 60% de dichos lotes no contiene componentes defectuosos, el 30% contiene un componente defectuoso y el 10% contiene dos componentes defectuosos. Se elige un lote, se seleccionan y prueban al azar dos componentes del lote, y ninguno resulta defectuoso.   
PDF

a: ¿Cuál es la probabilidad de que no existan componentes defectuosos en el lote?   

b: ¿Cuál es la probabilidad de que exista un defectuoso en el lote?   

c: ¿Cuál es la probabilidad de que existan dos defectuosos en el lote?

```julia
using Printf

P_L0 = 0.60
P_L1 = 0.30
P_L2 = 0.10

P_T_given_L0 = 1.0
P_T_given_L1 = binomial(19, 2) / binomial(20, 2)
P_T_given_L2 = binomial(18, 2) / binomial(20, 2)

P_T = (P_T_given_L0 * P_L0) + (P_T_given_L1 * P_L1) + (P_T_given_L2 * P_L2)

prob_a = (P_T_given_L0 * P_L0) / P_T
prob_b = (P_T_given_L1 * P_L1) / P_T
prob_c = (P_T_given_L2 * P_L2) / P_T

@printf("(a) La probabilidad de que no existan componentes defectuosos en el lote es de %.3f (%.1f%%).\n", prob_a, prob_a * 100)
@printf("(b) La probabilidad de que exista un defectuoso en el lote es de %.3f (%.1f%%).\n", prob_b, prob_b * 100)
@printf("(c) La probabilidad de que existan dos defectuosos en el lote es de %.3f (%.1f%%).\n", prob_c, prob_c * 100)
```
