15. Demuestra que cuando todos los pesos son iguales, la media ponderada es igual a la media aritmética.

```julia
x15 = [5, 10, 15, 20]
w_iguales = [1, 1, 1, 1]

m_ponderada = mean(x15, weights(w_iguales))
m_aritmetica = mean(x15)

println("Media Ponderada: $m_ponderada | Media Aritmética: $m_aritmetica")

println("• Al asignar el mismo peso (w = 1) a todos los elementos, la contribución de cada dato es idéntica.")
println("• Matemáticamente, Σ(w_i * x_i) / Σ(w_i) se reduce a Σ(x_i) / N, demostrando que la media ponderada es equivalente a la media aritmética simple.")
```

20. Verifica que std(x)^2 == var(x) para un conjunto de datos x dado.

```julia
x20 = [2, 4, 6, 8, 10]
std_cuadrado = std(x20)^2
varianza = var(x20)

println("std(x)^2: $std_cuadrado | var(x): $varianza")

println("• Verificación exitosa: std(x)^2 == var(x) evalúa a $(std_cuadrado == varianza).")
println("• Por definición, la desviación estándar es la raíz cuadrada de la varianza; elevarla al cuadrado retorna exactamente la varianza muestral.")
```

24. Simula 100 lanzamientos de dado, calcula las frecuencias relativas y compáralas con la probabilidad teórica de 1/6.

```julia
x24 = rand(1:6, 100)
f_relativas = countmap(x24)
prob_teorica = 1 / 6

p24 = bar(1:6, [get(f_relativas, i, 0)/100 for i in 1:6], label="Frecuencia Relativa", title="Punto 24: Frecuencias Relativas vs Teórico", xlabel="Cara", ylabel="Proporción", color=:teal)
hline!(p24, [prob_teorica], label="Teórico (1/6 ≈ 0.167)", color=:red, lw=2)

display(p24)

println("• La probabilidad teórica de cada cara es 1/6 ≈ 0.1667 (16.67%).")
println("• En la simulación de 100 lanzamientos, las frecuencias observadas oscilan alrededor de esta línea teórica debido a la ley de los grandes números.")
```

29. Genera 100 valores aleatorios de una distribución normal y calcula asimetría y curtosis. ¿Qué observas?

```julia
x29 = rand(Normal(0, 1), 100)
sk29 = skewness(x29)
kt29 = kurtosis(x29)

println("Skewness: $(round(sk29, digits=3)) | Kurtosis: $(round(kt29, digits=3))")

println("• Asimetría (Skewness = $(round(sk29, digits=2))): Cercana a 0, indicando una distribución prácticamente simétrica alrededor de la media.")
println("• Curtosis ($(round(kt29, digits=2))): Cercana a 0 (exceso de curtosis en StatsBase), confirmando la forma mesocúrtica propia de la distribución Normal.")
```

30. Compara la asimetría y curtosis de x1 = rand(Normal(0,1), 1000) y x2 = rand(Exponential(1), 1000).

```julia
x1_norm = rand(Normal(0, 1), 1000)
x2_exp = rand(Exponential(1), 1000)

sk_norm, kt_norm = skewness(x1_norm), kurtosis(x1_norm)
sk_exp, kt_exp = skewness(x2_exp), kurtosis(x2_exp)

println("Normal     -> Skewness: $(round(sk_norm, digits=2)), Kurtosis: $(round(kt_norm, digits=2))")
println("Exponencial -> Skewness: $(round(sk_exp, digits=2)), Kurtosis: $(round(kt_exp, digits=2))")

println("• Distribución Normal: Presenta asimetría cercana a 0 (simétrica) y curtosis cercana a 0 (mesocúrtica).")
println("• Distribución Exponencial: Presenta una asimetría altamente positiva (sesgada a la derecha) y una curtosis elevada (leptocúrtica), debido a la concentración de datos cerca del cero y una cola larga a la derecha.")
```
